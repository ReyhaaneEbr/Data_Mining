This project demonstrates image compression using Singular Value Decomposition (SVD). Both grayscale and color images are compressed by retaining only the largest singular values.

## 1. Required Libraries

Import the libraries required for image processing, numerical computation, and visualization.

## 2. Helper Functions

- `show()`: Displays an image.
- `openImage()`: Loads an image and converts it into a NumPy array.
- `compress()`: Applies SVD and reconstructs the image using the first k singular values.

## 3. Grayscale Image Compression

A two-dimensional black-and-white image is selected and decomposed using SVD. The image is reconstructed using different values of k (4, 8, and 12). The original and compressed image sizes are then compared.

The approximate storage size of the compressed image is calculated as:

$$k(1+m+n)$$

where m and n represent the image dimensions.

## 4. Color Image Compression

A color image is represented by three separate channels: Red, Green, and Blue (RGB). SVD is applied independently to each channel. The compressed channels are then merged to reconstruct the color image.

The compression process is repeated for k = 5, 10, 20, 100, and 200.

## 5. Results and Comparison

The reconstructed images are displayed for different values of k. Smaller values of k require less storage but generally produce lower image quality. Larger values of k preserve more image details.

```python
# Import Libraries
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

# Display Image
def show(image, cmap=None):
    plt.imshow(image, cmap=cmap)
    plt.axis('off')
    plt.show()

# Open Image
def openImage(imagePath):
    imOrig = Image.open(imagePath)
    im = np.array(imOrig)
    return im

# SVD Image Compression
def compress(Matrix, k):
    u, s, v = np.linalg.svd(Matrix, full_matrices=False)

    leftSide = np.matmul(u[:, :k], np.diag(s[:k]))
    CompressedInner = np.matmul(leftSide, v[:k, :])

    Compressed = np.clip(CompressedInner, 0, 255).astype('uint8')
    return Compressed

# Grayscale Image Compression
imOrig = Image.open('/content/img_1.jpg').convert('L')
im = np.array(imOrig)
(m, n) = im.shape

print('Original Grayscale Image:')
show(im, cmap='gray')
print('Original size:', m * n)
print('********')

for k in (4, 8, 12):
    print('k =', k)
    originalSize = m * n
    print('Original size:', originalSize)

    compressImage = compress(im, k)
    newImage = Image.fromarray(compressImage)

    compressedSize = k * (1 + m + n)
    print('Compressed size:', compressedSize)
    show(newImage, cmap='gray')
    print('********')

# Color Image Compression
colorImage = Image.open('/content/img_2.jpg').convert('RGB')
colorArray = np.array(colorImage)
(m, n, channels) = colorArray.shape

print('Original Color Image:')
show(colorArray)
print('Original size:', m * n * channels)
print('********')

for k in (5, 10, 20, 100, 200):
    print('k =', k)

    # Separate RGB Channels
    ared = colorArray[:, :, 0]
    agreen = colorArray[:, :, 1]
    ablue = colorArray[:, :, 2]

    # Compress Each Channel Separately
    redCompressed = compress(ared, k)
    greenCompressed = compress(agreen, k)
    blueCompressed = compress(ablue, k)

    # Merge RGB Channels
    compressedColor = np.stack(
        (redCompressed, greenCompressed, blueCompressed), axis=2
    )

    newColorImage = Image.fromarray(compressedColor, mode='RGB')

    # Calculate Compressed Size for Three Channels
    originalSize = m * n * 3
    compressedSize = 3 * k * (1 + m + n)

    print('Original size:', originalSize)
    print('Compressed size:', compressedSize)
    show(newColorImage)
    print('********')
```
